```python
import pandas as pd

# 1. Carga el dataset 
df = pd.read_csv('../data/coah.csv')

print("--- INFORMACIÓN GENERAL DEL DATASET ---")
print(f"Total de registros (filas): {len(df)}")
print(f"Columnas disponibles: {list(df.columns)}\n")

# 2. Verificar valores faltantes o nulos
print("--- VALORES FALTANTES ---")
print(df.isnull().sum(), "\n")

# 3. Distribución de clases / etiquetas (usando 'sentimiento')
print("--- DISTRIBUCIÓN DE CLASES (SENTIMIENTO) ---")
print(df['sentimiento'].value_counts())

print("\n" + "="*50 + "\n")

# 4. Inspección visual aleatoria
print("--- MUESTRA ALEATORIA DE 5 RESEÑAS ---")
muestra = df.sample(5, random_state=42)


for index, row in muestra.iterrows():
    print(f"ID: {row['id']} | Rank: {row['rank']}")
    print(f"Título: {row['titulo']}")
    print(f"Texto: {row['texto']}") 
    print(f"Sentimiento: {row['sentimiento']}")
    print("-" * 40)
```

```python
# 5. Duplicados (solo detección) ---
print("Textos duplicados:", df["texto"].duplicated().sum())
rep = df[df["texto"].duplicated(keep=False)]
print("Filas implicadas:", len(rep))
print("Pares con puntuación distinta:", (rep.groupby("texto")["rank"].nunique() > 1).sum())
```

```python
# 6.Longitud de los textos ---
df["palabras"] = df["texto"].str.split().str.len()
print(df["palabras"].describe().round(1))
print(df.groupby("sentimiento")["palabras"].mean().round(1))
print("Textos con menos de 20 palabras:", (df["palabras"] < 20).sum())
```

```python
#7.Palabras más frecuentes por clase ---
import re
from collections import Counter
orden = ["negativo", "neutro", "positivo"]

parar = {"de","la","el","en","y","a","que","los","se","las","un","una","con","por",
         "del","para","es","me","al","lo","su","muy","más","como","pero","le","mi","nos","fue"}

def top(serie, n=10):
    c = Counter(w for t in serie for w in re.findall(r"[a-záéíóúñü]+", t.lower())
                if w not in parar and len(w) > 2)
    return c.most_common(n)

for clase in orden:
    print(clase, top(df.loc[df["sentimiento"] == clase, "texto"]))
```
